### Lab Exercise 3.1: Creating a List and Accessing Items by Index

**Context**

A **list** lets you store multiple related values under one variable name. Instead of writing `bike1 = 'trek'`, `bike2 = 'redline'`, `bike3 = 'specialized'` as three separate variables, you group them: `bicycles = ['trek', 'redline', 'specialized']`. This is how Python organises collections — a playlist of songs, a list of usernames, a shopping cart — all use this pattern.

Python uses **zero-based indexing**: the first item is at position 0, not 1. This feels odd at first, but it is universal across programming. Negative indices count from the end: `my_list[-1]` always gives the last item, no matter how long the list is.

**Concept Flow**

  1. `Square brackets `[]` hold the list; commas separate items`
  2. my_list[0]  -->  first item   (index starts at 0)
  3. my_list[1]  -->  second item
  4. my_list[-1] -->  last item    (negative counts from end)
  5. my_list[-2] -->  second-to-last item
  6. Python's .title() / .upper() work on list items just like on plain strings

**Code Preview**

```python
# Adapted from source example 3.1
bicycles = ['trek', 'cannondale', 'redline', 'specialized']
message = f"My first bicycle was a {bicycles[0].title()}."
print(message)
```

````python
# Adapted from source example 3.1
bicycles = ['trek', 'cannondale', 'redline', 'specialized']
message = f"My first bicycle was a {bicycles[0].title()}."

print(message)
````

### Lab Exercise 3.2: Reversing a List with .reverse()

**Context**

`.reverse()` flips the order of items in a list permanently — the last item becomes first, and so on. It is not alphabetical reversal; it simply mirrors whatever order the list currently has. Call `.reverse()` again to restore the original order (since reversing a reversed list gives you back where you started).

**Concept Flow**

  1. my_list.reverse()  -->  flips the list order in place
  2. Not alphabetical: it reverses the current order, whatever that is
  3. Permanent but undoable: call .reverse() again to get the original back
  4. Useful for displaying items in chronological reverse (newest first)

**Code Preview**

```python
# Adapted from source example 3.2
cars = ['bmw', 'audi', 'toyota', 'subaru']
print(cars)
cars.reverse()
print(cars)
```

### Lab Exercise 3.4: Creating a List and Accessing Items by Index

**Context**

A **list** lets you store multiple related values under one variable name. Instead of writing `bike1 = 'trek'`, `bike2 = 'redline'`, `bike3 = 'specialized'` as three separate variables, you group them: `bicycles = ['trek', 'redline', 'specialized']`. This is how Python organises collections — a playlist of songs, a list of usernames, a shopping cart — all use this pattern.

Python uses **zero-based indexing**: the first item is at position 0, not 1. This feels odd at first, but it is universal across programming. Negative indices count from the end: `my_list[-1]` always gives the last item, no matter how long the list is.

**Concept Flow**

  1. `Square brackets `[]` hold the list; commas separate items`
  2. my_list[0]  -->  first item   (index starts at 0)
  3. my_list[1]  -->  second item
  4. my_list[-1] -->  last item    (negative counts from end)
  5. my_list[-2] -->  second-to-last item
  6. Python's .title() / .upper() work on list items just like on plain strings

**Code Preview**

```python
# Adapted from source example 3.4
bicycles = ['trek', 'cannondale', 'redline', 'specialized']
print(bicycles)
```

````python
# Adapted from source example 3.4
bicycles = ['trek', 'cannondale', 'redline', 'specialized']
print(bicycles)
````

### Lab Exercise 3.10: Sorting a List Permanently with .sort()

**Context**

`.sort()` rearranges the items of a list **in place** — alphabetically for strings, numerically for numbers. After calling `.sort()`, the original order is gone forever. Use it when you want to keep the sorted order as the list's permanent state.

To sort in reverse order, pass `reverse=True`. This is useful for displaying items from largest to smallest (highest scores, most recent dates).

**Concept Flow**

  1. my_list.sort()              -->  sorts in place, ascending (A→Z, 0→9)
  2. my_list.sort(reverse=True)  -->  sorts in place, descending (Z→A, 9→0)
  3. The list is modified permanently; the original order cannot be recovered
  4. Items must be comparable: all strings or all numbers (not mixed)

**Code Preview**

```python
# Adapted from source example 3.10
cars = ['bmw', 'audi', 'toyota', 'subaru']
cars.sort()
print(cars)
```

````python
# Adapted from source example 3.10
cars = ['bmw', 'audi', 'toyota', 'subaru']
cars.sort()
print(cars)
````

### Lab Exercise 3.12: Displaying a Sorted List Temporarily with sorted()

**Context**

`sorted()` is a **function** (not a method) that returns a *new* sorted list, leaving the original unchanged. Use it when you need to display items in alphabetical order for one moment without losing the original order — for example, showing a user a sorted preview while keeping internal order intact.

The distinction: `.sort()` modifies the list forever; `sorted()` creates a temporary copy. The book teaches this difference explicitly: always ask yourself whether you need the sorted order to persist.

**Concept Flow**

  1. sorted(my_list)             -->  returns a new sorted list; original unchanged
  2. sorted(my_list, reverse=True) -->  returns new list sorted descending
  3. The original list remains in its original order after sorted()
  4. You can print sorted() directly or assign it to a new variable

**Code Preview**

```python
# Adapted from source example 3.12
cars = ['bmw', 'audi', 'toyota', 'subaru']
print("Here is the original list:")
print(cars)
print("\nHere is the sorted list:")
print(sorted(cars))
print("\nHere is the original list again:")
print(cars)
```

````python
# Adapted from source example 3.12
cars = ['bmw', 'audi', 'toyota', 'subaru']

print("Here is the original list:")
print(cars)

print("\nHere is the sorted list:")
print(sorted(cars))

print("\nHere is the original list again:")
print(cars)
````

### Lab Exercise 3.17: Adding Items with .append()

**Context**

`.append()` adds one item to the **end** of a list. It is the most common way to build a list dynamically — you start with an empty list `[]` and then call `.append()` inside a loop or whenever new data arrives. Every social media post added to a feed, every task added to a to-do list, works exactly this way.

The book's habit to build early: *start with an empty list, then append items one by one*. This pattern scales to hundreds of items without any change to your code.

**Concept Flow**

  1. my_list.append('item')  -->  adds 'item' at the end
  2. The list grows by one element each time .append() is called
  3. Order is preserved: appended items stay in the order you added them
  4. Start with []  then .append()  -->  standard pattern for dynamic lists

**Code Preview**

```python
# Adapted from source example 3.17
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)
motorcycles.append('ducati')
print(motorcycles)
```

### Lab Exercise 3.19: Inserting Items at a Specific Position with .insert()

**Context**

While `.append()` always adds to the end, `.insert()` lets you place a new item *anywhere* in the list. The first argument is the index where the new item should go; all existing items from that position onward slide one step to the right.

This matters when order is meaningful — inserting a high-priority task at position 0, or adding a new chapter between existing chapters in a book index.

**Concept Flow**

  1. my_list.insert(0, 'value')  -->  inserts at the beginning
  2. my_list.insert(2, 'value')  -->  inserts before the current item at index 2
  3. Every item at index >= the insertion point shifts right by one
  4. List grows by one; no items are removed or overwritten

**Code Preview**

```python
# Adapted from source example 3.19
motorcycles = ['honda', 'yamaha', 'suzuki']
motorcycles.insert(0, 'ducati')
print(motorcycles)
```

### Lab Exercise 3.20: Removing Items Permanently with del

**Context**

The `del` statement removes an item by index with no return value — the item is simply gone. Use `del` when you are certain you no longer need the value at all. Unlike `.pop()`, there is nothing to capture; unlike `.remove()`, you specify the position, not the value.

**Concept Flow**

  1. del my_list[0]   -->  permanently removes item at index 0
  2. del my_list[-1]  -->  permanently removes the last item
  3. After del, all items after the deleted position shift left
  4. del also works on any variable: del my_var removes the variable entirely

**Code Preview**

```python
print(motorcycles)
```

### Lab Exercise 3.22: Removing Items with .pop()

**Context**

`.pop()` removes an item *and returns it* so you can use it. Called with no argument, it removes the last item. Called with an index, it removes that specific item. Think of a stack of plates: `.pop()` takes the top plate off and hands it to you.

The key difference from `.remove()`: `.pop()` works by **position**, not value, and it gives you back the removed item — which is useful when you need to process or display the item after removing it.

**Concept Flow**

  1. my_list.pop()      -->  removes AND returns the last item
  2. my_list.pop(i)     -->  removes AND returns the item at index i
  3. The returned value can be stored: removed = my_list.pop()
  4. Use .pop() when you want to use the item after removing it
  5. Use del when you just want to discard it

**Code Preview**

```python
# Adapted from source example 3.22
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)
popped_motorcycle = motorcycles.pop()
print(motorcycles)
print(popped_motorcycle)
```

````python
# Adapted from source example 3.22
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)

popped_motorcycle = motorcycles.pop()
print(motorcycles)
print(popped_motorcycle)
````

### Lab Exercise 4.1: Iterating Over a List with a for Loop

**Context**

A **for loop** lets Python visit every item in a list automatically, one by one, without you having to know or care how many items there are. This is the fundamental tool for processing collections: send an email to every user in a list, apply a discount to every price, print a greeting for every name.

The loop variable (e.g. `magician` in `for magician in magicians`) takes on each item's value in turn. The indented block runs once per item. Python knows the loop is over when items run out; any unindented code after the loop runs once, at the end.

**Concept Flow**

  1. for item in my_list:   -->  item takes each value in turn
  2. Indented block runs once per item: do anything you like here
  3. Loop ends automatically when the list is exhausted
  4. Code AFTER the loop (unindented) runs once, after all items are processed
  5. Choose a meaningful singular name: for car in cars, for name in names

**Code Preview**

```python
# Adapted from source example 4.1
dimensions = (200, 50)
print("Original dimensions:")
for dimension in dimensions:
    print(dimension)
dimensions = (400, 100)
print("\nModified dimensions:")
for dimension in dimensions:
    print(dimension)
```

````python
# Adapted from source example 4.1
dimensions = (200, 50)
print("Original dimensions:")
for dimension in dimensions:
    print(dimension)

dimensions = (400, 100)
print("\nModified dimensions:")
for dimension in dimensions:
    print(dimension)
````

### Lab Exercise 4.2: Generating Number Sequences with range()

**Context**

`range()` generates a sequence of numbers without storing them all in memory. `range(1, 6)` produces 1, 2, 3, 4, 5 — the stop value is *excluded*, which surprises beginners but is consistent with Python's general rule (slices work the same way). `range(n)` starts at 0 by default.

You use `range()` constantly: counting iterations, generating indices, creating lists of numbers with `list(range(...))`.

**Concept Flow**

  1. range(stop)        -->  0, 1, 2, ..., stop-1
  2. range(start, stop) -->  start, start+1, ..., stop-1  (stop NOT included)
  3. range(start, stop, step) -->  every step-th value
  4. list(range(1,6))   -->  [1, 2, 3, 4, 5]  (materialise as a list)

**Code Preview**

```python
# Adapted from source example 4.2
even_numbers = list(range(2, 11, 2))
print(even_numbers)
```

````python
# Adapted from source example 4.2
even_numbers = list(range(2, 11, 2))
print(even_numbers)
````

### Lab Exercise 4.13: Generating Number Sequences with range()

**Context**

`range()` generates a sequence of numbers without storing them all in memory. `range(1, 6)` produces 1, 2, 3, 4, 5 — the stop value is *excluded*, which surprises beginners but is consistent with Python's general rule (slices work the same way). `range(n)` starts at 0 by default.

You use `range()` constantly: counting iterations, generating indices, creating lists of numbers with `list(range(...))`.

**Concept Flow**

  1. range(stop)        -->  0, 1, 2, ..., stop-1
  2. range(start, stop) -->  start, start+1, ..., stop-1  (stop NOT included)
  3. range(start, stop, step) -->  every step-th value
  4. list(range(1,6))   -->  [1, 2, 3, 4, 5]  (materialise as a list)

**Code Preview**

```python
# Adapted from source example 4.13
even_numbers = list(range(2, 11, 2))
print(even_numbers)
```

````python
# Adapted from source example 4.13
even_numbers = list(range(2, 11, 2))
print(even_numbers)
````

### Lab Exercise 4.35: Generating Number Sequences with range()

**Context**

`range()` generates a sequence of numbers without storing them all in memory. `range(1, 6)` produces 1, 2, 3, 4, 5 — the stop value is *excluded*, which surprises beginners but is consistent with Python's general rule (slices work the same way). `range(n)` starts at 0 by default.

You use `range()` constantly: counting iterations, generating indices, creating lists of numbers with `list(range(...))`.

**Concept Flow**

  1. range(stop)        -->  0, 1, 2, ..., stop-1
  2. range(start, stop) -->  start, start+1, ..., stop-1  (stop NOT included)
  3. range(start, stop, step) -->  every step-th value
  4. list(range(1,6))   -->  [1, 2, 3, 4, 5]  (materialise as a list)

**Code Preview**

```python
# Adapted from source example 4.35
squares = []
for value in range(1, 11):
    square = value ** 2
    squares.append(square)
print(squares)
```

````python
# Adapted from source example 4.35
squares = []
for value in range(1, 11):
    square = value ** 2
    squares.append(square)

print(squares)
````

### Lab Exercise 5.1: Making Decisions with if Statements

**Context**

An `if` statement runs a block of code only when a condition is `True`. If the condition is `False`, the block is skipped entirely. This is the most fundamental decision-making tool in programming: validate input, check permissions, respond differently to different data.

The indented block can contain any code — multiple lines, other if statements, function calls. Python uses indentation (not braces) to define the block.

**Concept Flow**

  1. if condition:      -->  runs indented block only when condition is True
  2. if age >= 18:      -->  real example: age gate
  3. Multiple indented lines all run if the condition is True
  4. If False: the entire indented block is skipped; execution continues after

**Code Preview**

```python
# Adapted from source example 5.1
age = 12
if age < 4:
    price = 0
elif age < 18:
    price = 25
elif age < 65:
    price = 40
elif age >= 65:
    price = 20
print(f"Your admission cost is ${price}.")
```

````python
# Adapted from source example 5.1
age = 12

if age < 4:
    price = 0
elif age < 18:
    price = 25
elif age < 65:
    price = 40
elif age >= 65:
    price = 20

print(f"Your admission cost is ${price}.")
````

### Lab Exercise 5.2: Checking Membership with in and not in

**Context**

The `in` keyword checks whether a value exists in a list (or string) and returns `True` or `False`. `not in` is the opposite. This is one of Python's most readable features — the code almost reads like English: `if 'mushrooms' in requested_toppings:`

Real uses: checking if a username is banned, if a product is in stock, if a required field is in a form submission.

**Concept Flow**

  1. value in my_list      -->  True if value is found anywhere in the list
  2. value not in my_list  -->  True if value is NOT in the list
  3. Works on strings too: 'py' in 'python'  -->  True
  4. Case-sensitive: 'Python' in ['python'] is False

**Code Preview**

```python
# Adapted from source example 5.2
banned_users = ['andrew', 'carolina', 'david']
user = 'marie'
if user not in banned_users:
    print(f"{user.title()}, you can post a response if you wish.")
```

````python
# Adapted from source example 5.2
banned_users = ['andrew', 'carolina', 'david']
user = 'marie'

if user not in banned_users:
    print(f"{user.title()}, you can post a response if you wish.")
````

### Lab Exercise 5.5: Using if Statements Inside Loops Over Lists

**Context**

Combining a for loop with an if statement lets you treat certain items differently from others. The book's pizzeria example: loop over requested toppings, but if a topping is unavailable, print a "sorry" message instead of "adding". This pattern — loop + condition + different action — is fundamental to data processing and filtering.

**Concept Flow**

  1. for item in my_list:              -->  visit every item
  2.     if item == 'special_value':   -->  single out one item
  3.         do something different
  4.     else:                         -->  handle all other items
  5.         do the normal thing
  6. Combine with 'in' to check against an availability list

**Code Preview**

```python
# Adapted from source example 5.5
available_toppings = ['mushrooms', 'olives', 'green peppers',
                        'pepperoni', 'pineapple', 'extra cheese']
requested_toppings = ['mushrooms', 'french fries', 'extra cheese']
for requested_topping in requested_toppings:
    if requested_topping in available_toppings:
        print(f"Adding {requested_topping}.")
    else:
        print(f"Sorry, we don't have {requested_topping}.")
print("\nFinished making your pizza!")
```

````python
# Adapted from source example 5.5
available_toppings = ['mushrooms', 'olives', 'green peppers',
                        'pepperoni', 'pineapple', 'extra cheese']

requested_toppings = ['mushrooms', 'french fries', 'extra cheese']

for requested_topping in requested_toppings:
    if requested_topping in available_toppings:
        print(f"Adding {requested_topping}.")
    else:
        print(f"Sorry, we don't have {requested_topping}.")

print("\nFinished making your pizza!")
````

### Lab Exercise 5.11: Checking Membership with in and not in

**Context**

The `in` keyword checks whether a value exists in a list (or string) and returns `True` or `False`. `not in` is the opposite. This is one of Python's most readable features — the code almost reads like English: `if 'mushrooms' in requested_toppings:`

Real uses: checking if a username is banned, if a product is in stock, if a required field is in a form submission.

**Concept Flow**

  1. value in my_list      -->  True if value is found anywhere in the list
  2. value not in my_list  -->  True if value is NOT in the list
  3. Works on strings too: 'py' in 'python'  -->  True
  4. Case-sensitive: 'Python' in ['python'] is False

**Code Preview**

```python
# Adapted from source example 5.11
banned_users = ['andrew', 'carolina', 'david']
user = 'marie'
if user not in banned_users:
    print(f"{user.title()}, you can post a response if you wish.")
```

````python
# Adapted from source example 5.11
banned_users = ['andrew', 'carolina', 'david']
user = 'marie'

if user not in banned_users:
    print(f"{user.title()}, you can post a response if you wish.")
````

### Lab Exercise 5.16: Writing Conditional Tests

**Context**

A **conditional test** is any expression that evaluates to `True` or `False`. Python uses `==` (equality), `!=` (not equal), `<`, `>`, `<=`, `>=` to compare values. You use these constantly: checking whether a user is logged in, whether a price is within budget, whether a username is already taken.

Case matters for strings: `'BMW' == 'bmw'` is `False`. Use `.lower()` on both sides for case-insensitive comparison — a technique the book calls out specifically.

**Concept Flow**

  1. ==  checks equality   (= assigns, == compares)
  2. !=  checks inequality
  3. >  >=  <  <=  for numerical comparisons
  4. 'BMW'.lower() == 'bmw'  -->  True  (case-insensitive pattern)
  5. Conditional test result is always True or False (Boolean)

**Code Preview**

```python
# Adapted from source example 5.16
requested_toppings = ['mushrooms', 'green peppers', 'extra cheese']
for requested_topping in requested_toppings:
    print(f"Adding {requested_topping}.")
print("\nFinished making your pizza!")
```

````python
# Adapted from source example 5.16
requested_toppings = ['mushrooms', 'green peppers', 'extra cheese']

for requested_topping in requested_toppings:
    print(f"Adding {requested_topping}.")

print("\nFinished making your pizza!")
````

```python
# Adapted from source example 5.19
available_toppings = ['mushrooms', 'olives', 'green peppers',
                        'pepperoni', 'pineapple', 'extra cheese']

requested_toppings = ['mushrooms', 'french fries', 'extra cheese']

for requested_topping in requested_toppings:
    if requested_topping in available_toppings:
        print(f"Adding {requested_topping}.")
    else:
        print(f"Sorry, we don't have {requested_topping}.")

print("\nFinished making your pizza!")
```

### Lab Exercise 5.20: Making Decisions with if Statements

**Context**

An `if` statement runs a block of code only when a condition is `True`. If the condition is `False`, the block is skipped entirely. This is the most fundamental decision-making tool in programming: validate input, check permissions, respond differently to different data.

The indented block can contain any code — multiple lines, other if statements, function calls. Python uses indentation (not braces) to define the block.

**Concept Flow**

  1. if condition:      -->  runs indented block only when condition is True
  2. if age >= 18:      -->  real example: age gate
  3. Multiple indented lines all run if the condition is True
  4. If False: the entire indented block is skipped; execution continues after

**Code Preview**

```python
# Adapted from source example 5.20
age = 19
if age >= 18:
    print("You are old enough to vote!")
```

````python
# Adapted from source example 5.20
age = 19
if age >= 18:
    print("You are old enough to vote!")
````

### Lab Exercise 6.2: Safely Accessing Values with .get()

**Context**

`.get()` retrieves a value by key but returns a **default** instead of raising an error if the key doesn't exist. `my_dict.get('missing_key', 'N/A')` returns `'N/A'` rather than crashing. This is the safe alternative to direct bracket access whenever a key's presence is uncertain — for example, reading optional user settings or looking up scores for players who may not have played yet.

**Concept Flow**

  1. my_dict.get('key')           -->  returns value or None if missing
  2. my_dict.get('key', default)  -->  returns value or default if missing
  3. No KeyError raised; program continues normally
  4. Use when a missing key is an expected, normal case

**Code Preview**

```python
# Adapted from source example 6.2
alien_0 = {'color': 'green', 'speed': 'slow'}
point_value = alien_0.get('points', 'No point value assigned.')
print(point_value)
```

````python
# Adapted from source example 6.2
alien_0 = {'color': 'green', 'speed': 'slow'}

point_value = alien_0.get('points', 'No point value assigned.')
print(point_value)
````

### Lab Exercise 6.4: Looping Through a Dictionary

**Context**

Dictionaries support looping through their key-value pairs (`.items()`), keys only (`.keys()`), or values only (`.values()`). This lets you process every record: print every user's profile, look up every product's price, find all responses that match a criterion.

`.items()` returns each pair as `(key, value)` which you unpack with two loop variables. `.keys()` (the default) gives just the keys. `.values()` gives just the values.

**Concept Flow**

  1. for k, v in my_dict.items(): -->  loop over key-value pairs
  2. for k in my_dict.keys():    -->  loop over keys (also: for k in my_dict)
  3. for v in my_dict.values():  -->  loop over values
  4. sorted(my_dict.keys())      -->  iterate keys in alphabetical order

**Code Preview**

```python
# Adapted from source example 6.4
favorite_languages = {
      'jen': ['python', 'rust'],
      'sarah': ['c'],
      'edward': ['rust', 'go'],
      'phil': ['python', 'haskell'],
      }
for name, languages in favorite_languages.items():
    print(f"\n{name.title()}'s favorite languages are:")
    for language in languages:
        print(f"\t{language.title()}")
```

````python
# Adapted from source example 6.4
favorite_languages = {
      'jen': ['python', 'rust'],
      'sarah': ['c'],
      'edward': ['rust', 'go'],
      'phil': ['python', 'haskell'],
      }

for name, languages in favorite_languages.items():
    print(f"\n{name.title()}'s favorite languages are:")
    for language in languages:
        print(f"\t{language.title()}")
````

### Lab Exercise 6.17: Safely Accessing Values with .get()

**Context**

`.get()` retrieves a value by key but returns a **default** instead of raising an error if the key doesn't exist. `my_dict.get('missing_key', 'N/A')` returns `'N/A'` rather than crashing. This is the safe alternative to direct bracket access whenever a key's presence is uncertain — for example, reading optional user settings or looking up scores for players who may not have played yet.

**Concept Flow**

  1. my_dict.get('key')           -->  returns value or None if missing
  2. my_dict.get('key', default)  -->  returns value or default if missing
  3. No KeyError raised; program continues normally
  4. Use when a missing key is an expected, normal case

**Code Preview**

```python
# Adapted from source example 6.17
alien_0 = {'color': 'green', 'speed': 'slow'}
point_value = alien_0.get('points', 'No point value assigned.')
print(point_value)
```

````python
# Adapted from source example 6.17
alien_0 = {'color': 'green', 'speed': 'slow'}

point_value = alien_0.get('points', 'No point value assigned.')
print(point_value)
````

### Lab Exercise 6.23: Looping Through a Dictionary

**Context**

Dictionaries support looping through their key-value pairs (`.items()`), keys only (`.keys()`), or values only (`.values()`). This lets you process every record: print every user's profile, look up every product's price, find all responses that match a criterion.

`.items()` returns each pair as `(key, value)` which you unpack with two loop variables. `.keys()` (the default) gives just the keys. `.values()` gives just the values.

**Concept Flow**

  1. for k, v in my_dict.items(): -->  loop over key-value pairs
  2. for k in my_dict.keys():    -->  loop over keys (also: for k in my_dict)
  3. for v in my_dict.values():  -->  loop over values
  4. sorted(my_dict.keys())      -->  iterate keys in alphabetical order

**Code Preview**

```python
# Adapted from source example 6.23
favorite_languages = {
    'jen': 'python',
    'sarah': 'c',
    'edward': 'rust',
    'phil': 'python',
    }
for name, language in favorite_languages.items():
    print(f"{name.title()}'s favorite language is {language.title()}.")
```

````python
# Adapted from source example 6.23
favorite_languages = {
    'jen': 'python',
    'sarah': 'c',
    'edward': 'rust',
    'phil': 'python',
    }

for name, language in favorite_languages.items():
    print(f"{name.title()}'s favorite language is {language.title()}.")
````

Use dictionaries to model real-world objects: a user profile, a product, a configuration setting, a row in a database.

Python creates dictionaries with curly braces `{}` and colons separating keys from values.

**Concept Flow**

  1. my_dict = {'key': 'value', 'key2': 'value2'}
  2. my_dict['key']  -->  returns the value associated with 'key'
  3. Keys must be unique; values can be anything
  4. Access a key that doesn't exist raises KeyError; use .get() to be safe

### Lab Exercise 7.1: Exiting a Loop Immediately with break

**Context**

`break` exits a while (or for) loop *immediately*, regardless of the condition. It is used when you have detected a "stop" signal mid-loop: the user typed 'quit', an item was found in a search, a limit was reached. The loop ends right there; any remaining code in the loop body is skipped for that iteration.

**Concept Flow**

  1. while True:         -->  intentionally infinite — relies on break to exit
  2.     ...             -->  do work
  3.     if quit_signal: break  -->  exits the loop immediately
  4. break skips the rest of the loop body AND the condition check
  5. Code after the while block resumes normally after break

**Code Preview**

```python
prompt = "\nPlease enter the name of a city you have visited:"
prompt += "\n(Enter 'quit' when you are finished.) "
while True:
    city = input(prompt)
    if city == 'quit':
        break
    else:
        print(f"I'd love to go to {city.title()}!")
```

````python
prompt = "\nPlease enter the name of a city you have visited:"
prompt += "\n(Enter 'quit' when you are finished.) "

while True:
    city = input(prompt)

    if city == 'quit':
        break
    else:
        print(f"I'd love to go to {city.title()}!")
````

### Lab Exercise 7.2: Using while Loops to Process and Move Items Between Lists

**Context**

A while loop with `.pop()` is the standard pattern for draining one list into another while processing each item. This simulates real workflows: move messages from "unprocessed" to "sent", move items from "cart" to "purchased", move jobs from "queue" to "completed". The loop ends naturally when the source list is empty.

**Concept Flow**

  1. while source_list:              -->  True while source_list is non-empty
  2.     item = source_list.pop()    -->  remove last item and capture it
  3.     process(item)               -->  do something with it
  4.     destination_list.append(item)  -->  move it to the other list
  5. After loop: source_list is [], destination_list has all items

**Code Preview**

```python
# Adapted from source example 7.2
# Start with users that need to be verified,
# and an empty list to hold confirmed users.
unconfirmed_users = ['alice', 'brian', 'candace']
confirmed_users = []
# Verify each user until there are no more unconfirmed users.
# Move each verified user into the list of confirmed users.
while unconfirmed_users:
    current_user = unconfirmed_users.pop()
    print(f"Verifying user: {current_user.title()}")
    confirmed_users.append(current_user)
# Display all confirmed users.
print("\nThe following users have been confirmed:")
for confirmed_user in confirmed_users:
    print(confirmed_user.title())
```

````python
# Adapted from source example 7.2
# Start with users that need to be verified,
# and an empty list to hold confirmed users.
unconfirmed_users = ['alice', 'brian', 'candace']
confirmed_users = []

# Verify each user until there are no more unconfirmed users.
# Move each verified user into the list of confirmed users.
while unconfirmed_users:
    current_user = unconfirmed_users.pop()
    print(f"Verifying user: {current_user.title()}")
    confirmed_users.append(current_user)

# Display all confirmed users.
print("\nThe following users have been confirmed:")
for confirmed_user in confirmed_users:
    print(confirmed_user.title())
````

### Lab Exercise 7.3: Skipping Iterations with continue

**Context**

`continue` skips the *rest of the current iteration* and jumps back to the top of the loop to check the condition again. Use it to skip items you don't want to process: skip even numbers to print only odd ones, skip banned users, skip malformed records. Unlike `break` which ends the loop, `continue` just skips *this one* iteration.

**Concept Flow**

  1. continue  -->  skip rest of current iteration, go back to condition check
  2. break     -->  exit the loop entirely
  3. Common pattern: if unwanted_condition: continue
  4. Everything before 'continue' in the body still ran; everything after is skipped

**Code Preview**

```python
# Adapted from source example 7.3
current_number = 0
while current_number < 10:
    current_number += 1
    if current_number % 2 == 0:
        continue
    print(current_number)
```

````python
# Adapted from source example 7.3
current_number = 0
while current_number < 10:
    current_number += 1
    if current_number % 2 == 0:
        continue

    print(current_number)
````

### Lab Exercise 7.4: Using the Modulo Operator % to Find Remainders

**Context**

The **modulo operator** `%` returns the remainder after division. `10 % 3` is `1` because 10 divided by 3 leaves a remainder of 1. Its most common use: `n % 2 == 0` tells you whether `n` is even (no remainder when divided by 2). Real uses: alternating row colours in a table, distributing tasks in round-robin, deciding when to flush a batch.

**Concept Flow**

  1. 7 % 2   -->  1  (7 divided by 2 is 3 remainder 1)
  2. 6 % 2   -->  0  (6 divided by 2 is 3 remainder 0 -- even!)
  3. n % 2 == 0  -->  True means n is even
  4. n % 10 -->  last digit of n
  5. n % k == 0  -->  n is divisible by k

**Code Preview**

```python
number = input("Enter a number, and I'll tell you if it's even or odd: ")
number = int(number)
if number % 2 == 0:
    print(f"\nThe number {number} is even.")
else:
    print(f"\nThe number {number} is odd.")
```

````python
number = input("Enter a number, and I'll tell you if it's even or odd: ")
number = int(number)

if number % 2 == 0:
    print(f"\nThe number {number} is even.")
else:
    print(f"\nThe number {number} is odd.")
````

### Lab Exercise 7.6: Using a Flag Variable to Control a while Loop

**Context**

A **flag** is a Boolean variable (usually `active = True`) that acts as a signal. The while loop condition is simply `while active:`. When you want to stop the loop, set `active = False`. This is cleaner than putting complex conditions in the `while` line itself — all the logic for stopping is inside the loop where it's easy to read and modify.

**Concept Flow**

  1. active = True   -->  flag starts as True
  2. while active:   -->  loop continues while flag is True
  3. Inside loop: if some_condition: active = False  -->  schedules exit
  4. Loop checks flag again at the top; stops cleanly after current iteration

**Code Preview**

```python
responses = {}
# Set a flag to indicate that polling is active.
polling_active = True
while polling_active:
    # Prompt for the person's name and response.
    name = input("\nWhat is your name? ")
    response = input("Which mountain would you like to climb someday? ")
    # Store the response in the dictionary.
    responses[name] = response
    # Find out if anyone else is going to take the poll.
    repeat = input("Would you like to let another person respond? (yes/ no) ")
# ... 7 more line(s) in this source example
```

````python
responses = {}
# Set a flag to indicate that polling is active.
polling_active = True

while polling_active:
    # Prompt for the person's name and response.
    name = input("\nWhat is your name? ")
    response = input("Which mountain would you like to climb someday? ")

    # Store the response in the dictionary.
    responses[name] = response

    # Find out if anyone else is going to take the poll.
    repeat = input("Would you like to let another person respond? (yes/ no) ")
# ... 7 more line(s) in this source example
````

### Lab Exercise 7.7: Repeating Actions with while Loops

**Context**

A **while loop** keeps running its indented block as long as a condition stays `True`. Unlike a `for` loop that iterates a fixed collection, a `while` loop continues indefinitely until something changes the condition. Use it when you don't know in advance how many iterations you need: keep asking for input until the user types 'quit', keep trying a network request until it succeeds.

Always ensure something inside the loop eventually makes the condition `False`, otherwise you have an infinite loop.

**Concept Flow**

  1. while condition:   -->  runs block repeatedly while condition is True
  2. Something inside the loop MUST eventually make condition False
  3. current_number += 1  -->  typical counter increment
  4. while True: ... break  -->  alternative pattern (see next topic)

**Code Preview**

```python
    message = input(prompt)
    if message == 'quit':
        active = False
    else:
        print(message)
```

````python
prompt = "\nTell me something, and I will repeat it back to you:"
prompt += "\nEnter 'quit' to end the program. "

active = True
while active:
    message = input(prompt)

    if message == 'quit':
        active = False
    else:
        print(message)
````

### Lab Exercise 7.10: Exiting a Loop Immediately with break

**Context**

`break` exits a while (or for) loop *immediately*, regardless of the condition. It is used when you have detected a "stop" signal mid-loop: the user typed 'quit', an item was found in a search, a limit was reached. The loop ends right there; any remaining code in the loop body is skipped for that iteration.

**Concept Flow**

  1. while True:         -->  intentionally infinite — relies on break to exit
  2.     ...             -->  do work
  3.     if quit_signal: break  -->  exits the loop immediately
  4. break skips the rest of the loop body AND the condition check
  5. Code after the while block resumes normally after break

**Code Preview**

```python
prompt = "\nPlease enter the name of a city you have visited:"
prompt += "\n(Enter 'quit' when you are finished.) "
while True:
    city = input(prompt)
    if city == 'quit':
        break
    else:
        print(f"I'd love to go to {city.title()}!")
```

````python
prompt = "\nPlease enter the name of a city you have visited:"
prompt += "\n(Enter 'quit' when you are finished.) "

while True:
    city = input(prompt)

    if city == 'quit':
        break
    else:
        print(f"I'd love to go to {city.title()}!")
````

### Lab Exercise 7.11: Using while Loops to Process and Move Items Between Lists

**Context**

A while loop with `.pop()` is the standard pattern for draining one list into another while processing each item. This simulates real workflows: move messages from "unprocessed" to "sent", move items from "cart" to "purchased", move jobs from "queue" to "completed". The loop ends naturally when the source list is empty.

**Concept Flow**

  1. while source_list:              -->  True while source_list is non-empty
  2.     item = source_list.pop()    -->  remove last item and capture it
  3.     process(item)               -->  do something with it
  4.     destination_list.append(item)  -->  move it to the other list
  5. After loop: source_list is [], destination_list has all items

**Code Preview**

```python
# Adapted from source example 7.11
# Start with users that need to be verified,
# and an empty list to hold confirmed users.
unconfirmed_users = ['alice', 'brian', 'candace']
confirmed_users = []
# Verify each user until there are no more unconfirmed users.
# Move each verified user into the list of confirmed users.
while unconfirmed_users:
    current_user = unconfirmed_users.pop()
    print(f"Verifying user: {current_user.title()}")
    confirmed_users.append(current_user)
# Display all confirmed users.
print("\nThe following users have been confirmed:")
for confirmed_user in confirmed_users:
    print(confirmed_user.title())
```

````python
# Adapted from source example 7.11
# Start with users that need to be verified,
# and an empty list to hold confirmed users.
unconfirmed_users = ['alice', 'brian', 'candace']
confirmed_users = []

# Verify each user until there are no more unconfirmed users.
# Move each verified user into the list of confirmed users.
while unconfirmed_users:
    current_user = unconfirmed_users.pop()
    print(f"Verifying user: {current_user.title()}")
    confirmed_users.append(current_user)

# Display all confirmed users.
print("\nThe following users have been confirmed:")
for confirmed_user in confirmed_users:
    print(confirmed_user.title())
````

### Lab Exercise 7.13: Skipping Iterations with continue

**Context**

`continue` skips the *rest of the current iteration* and jumps back to the top of the loop to check the condition again. Use it to skip items you don't want to process: skip even numbers to print only odd ones, skip banned users, skip malformed records. Unlike `break` which ends the loop, `continue` just skips *this one* iteration.

**Concept Flow**

  1. continue  -->  skip rest of current iteration, go back to condition check
  2. break     -->  exit the loop entirely
  3. Common pattern: if unwanted_condition: continue
  4. Everything before 'continue' in the body still ran; everything after is skipped

**Code Preview**

```python
# Adapted from source example 7.13
current_number = 0
while current_number < 10:
    current_number += 1
    if current_number % 2 == 0:
        continue
    print(current_number)
```

````python
# Adapted from source example 7.13
current_number = 0
while current_number < 10:
    current_number += 1
    if current_number % 2 == 0:
        continue

    print(current_number)
````

### Lab Exercise 7.15: Using the Modulo Operator % to Find Remainders

**Context**

The **modulo operator** `%` returns the remainder after division. `10 % 3` is `1` because 10 divided by 3 leaves a remainder of 1. Its most common use: `n % 2 == 0` tells you whether `n` is even (no remainder when divided by 2). Real uses: alternating row colours in a table, distributing tasks in round-robin, deciding when to flush a batch.

**Concept Flow**

  1. 7 % 2   -->  1  (7 divided by 2 is 3 remainder 1)
  2. 6 % 2   -->  0  (6 divided by 2 is 3 remainder 0 -- even!)
  3. n % 2 == 0  -->  True means n is even
  4. n % 10 -->  last digit of n
  5. n % k == 0  -->  n is divisible by k

**Code Preview**

```python
number = input("Enter a number, and I'll tell you if it's even or odd: ")
number = int(number)
if number % 2 == 0:
    print(f"\nThe number {number} is even.")
else:
    print(f"\nThe number {number} is odd.")
```

````python
number = input("Enter a number, and I'll tell you if it's even or odd: ")
number = int(number)

if number % 2 == 0:
    print(f"\nThe number {number} is even.")
else:
    print(f"\nThe number {number} is odd.")
````

### Lab Exercise 7.18: Using a Flag Variable to Control a while Loop

**Context**

A **flag** is a Boolean variable (usually `active = True`) that acts as a signal. The while loop condition is simply `while active:`. When you want to stop the loop, set `active = False`. This is cleaner than putting complex conditions in the `while` line itself — all the logic for stopping is inside the loop where it's easy to read and modify.

**Concept Flow**

  1. active = True   -->  flag starts as True
  2. while active:   -->  loop continues while flag is True
  3. Inside loop: if some_condition: active = False  -->  schedules exit
  4. Loop checks flag again at the top; stops cleanly after current iteration

**Code Preview**

```python
responses = {}
# Set a flag to indicate that polling is active.
polling_active = True
while polling_active:
    # Prompt for the person's name and response.
    name = input("\nWhat is your name? ")
    response = input("Which mountain would you like to climb someday? ")
    # Store the response in the dictionary.
    responses[name] = response
    # Find out if anyone else is going to take the poll.
    repeat = input("Would you like to let another person respond? (yes/ no) ")
# ... 7 more line(s) in this source example
```

````python
responses = {}
# Set a flag to indicate that polling is active.
polling_active = True

while polling_active:
    # Prompt for the person's name and response.
    name = input("\nWhat is your name? ")
    response = input("Which mountain would you like to climb someday? ")

    # Store the response in the dictionary.
    responses[name] = response

    # Find out if anyone else is going to take the poll.
    repeat = input("Would you like to let another person respond? (yes/ no) ")
# ... 7 more line(s) in this source example
````

### Lab Exercise 7.23: Repeating Actions with while Loops

**Context**

A **while loop** keeps running its indented block as long as a condition stays `True`. Unlike a `for` loop that iterates a fixed collection, a `while` loop continues indefinitely until something changes the condition. Use it when you don't know in advance how many iterations you need: keep asking for input until the user types 'quit', keep trying a network request until it succeeds.

Always ensure something inside the loop eventually makes the condition `False`, otherwise you have an infinite loop.

**Concept Flow**

  1. while condition:   -->  runs block repeatedly while condition is True
  2. Something inside the loop MUST eventually make condition False
  3. current_number += 1  -->  typical counter increment
  4. while True: ... break  -->  alternative pattern (see next topic)

**Code Preview**

```python
# Adapted from source example 7.23
pets = ['dog', 'cat', 'dog', 'goldfish', 'cat', 'rabbit', 'cat']
print(pets)
while 'cat' in pets:
    pets.remove('cat')
print(pets)
```

````python
# Adapted from source example 7.23
pets = ['dog', 'cat', 'dog', 'goldfish', 'cat', 'rabbit', 'cat']
print(pets)

while 'cat' in pets:
    pets.remove('cat')

print(pets)
````

### Lab Exercise 7.24: Getting Text Input from the User with input()

**Context**

The argument to `input()` is the **prompt** — what the user sees. Write prompts that are clear and specific: tell the user exactly what format you expect.

The book's advice: if your prompt is long, store it in a variable and add a space before the closing quote so the user's answer starts cleanly after the prompt.

**Concept Flow**

  1. name = input('What is your name? ')   -->  waits for input, stores as string
  2. The prompt string is displayed; user types; Enter stores the result
  3. Always returns a string — convert with int() or float() for numbers
  4. Write prompts that tell users exactly what to type and in what format

**Code Preview**

### Lab Exercise 8.1: Returning Values from Functions

**Context**

A function can **return** a value to the calling code using `return`. The returned value can be stored in a variable, printed, or passed to another function. Return values let you separate *computation* (done inside the function) from *use* (done outside). This is the foundation of reusable, composable code.

**Concept Flow**

  1. return value   -->  sends value back to the caller
  2. result = f()  -->  captures the returned value
  3. print(f())    -->  uses the returned value directly
  4. Execution stops at return; code after return in the function doesn't run
  5. A function without return implicitly returns None

**Code Preview**

```python
# Adapted from source example 8.1
def get_formatted_name(first_name, last_name, middle_name=''):
    """Return a full name, neatly formatted."""
    if middle_name:
        full_name = f"{first_name} {middle_name} {last_name}"
    else:
        full_name = f"{first_name} {last_name}"
    return full_name.title()
musician = get_formatted_name('jimi', 'hendrix')
print(musician)
musician = get_formatted_name('john', 'hooker', 'lee')
print(musician)
```

````python
# Adapted from source example 8.1
def get_formatted_name(first_name, last_name, middle_name=''):
    """Return a full name, neatly formatted."""
    if middle_name:
        full_name = f"{first_name} {middle_name} {last_name}"
    else:
        full_name = f"{first_name} {last_name}"
    return full_name.title()

musician = get_formatted_name('jimi', 'hendrix')
print(musician)

musician = get_formatted_name('john', 'hooker', 'lee')
print(musician)
````

### Lab Exercise 8.3: Organising Functions in Modules and Importing Them

**Context**

A **module** is a `.py` file containing functions. Storing functions in modules means you can import them into any other program — your code becomes reusable across projects. `import module_name` imports the whole module; `from module_name import function` imports just one function. Aliases (`import module as m`) shorten long names.

**Concept Flow**

  1. import module_name               -->  access as module_name.function()
  2. from module import function      -->  call function() directly
  3. from module import function as f -->  alias: call f() instead
  4. import module as m               -->  alias the whole module: m.function()
  5. Keep import statements at the top of the file

**Code Preview**

### Lab Exercise 8.4: Defining and Calling Functions

**Context**

A **function** is a named block of code you define once and can run (call) as many times as you need. `def` announces the function definition; the indented block is the body; the **docstring** (triple-quoted string immediately after `def`) describes what the function does.

Functions are the first step toward reusable code: write the logic once, use it everywhere. They also make programs easier to read — a well-named function call like `greet_user('Jesse')` is self-documenting.

**Concept Flow**

  1. def function_name():   -->  defines the function
  2.     'Docstring'        -->  describes what it does (best practice)
  3.     # body             -->  code that runs when function is called
  4. function_name()       -->  calls (runs) the function
  5. Call the function as many times as needed, anywhere in the program

**Code Preview**

```python
# Adapted from source example 8.4
def make_pizza(size, *toppings):

    for topping in toppings:
        print(f"- {topping}")
```

A **function** is a named block of code you define once and can run (call) as many times as you need. `def` announces the function definition; the indented block is the body; the **docstring** (triple-quoted string immediately after `def`) describes what the function does.

Functions are the first step toward reusable code: write the logic once, use it everywhere. They also make programs easier to read — a well-named function call like `greet_user('Jesse')` is self-documenting.

**Concept Flow**

  1. def function_name():   -->  defines the function
  2.     'Docstring'        -->  describes what it does (best practice)
  3.     # body             -->  code that runs when function is called
  4. function_name()       -->  calls (runs) the function
  5. Call the function as many times as needed, anywhere in the program

**Code Preview**

```python
# Adapted from source example 8.14
def describe_pet(pet_name, animal_type='dog'):
    """Display information about a pet."""
    print(f"\nI have a {animal_type}.")
    print(f"My {animal_type}'s name is {pet_name.title()}.")
describe_pet(pet_name='willie')
```

````python
# Adapted from source example 8.14
def describe_pet(pet_name, animal_type='dog'):
    """Display information about a pet."""
    print(f"\nI have a {animal_type}.")
    print(f"My {animal_type}'s name is {pet_name.title()}.")

describe_pet(pet_name='willie')
````

### Lab Exercise 8.17: Returning Values from Functions

**Context**

A function can **return** a value to the calling code using `return`. The returned value can be stored in a variable, printed, or passed to another function. Return values let you separate *computation* (done inside the function) from *use* (done outside). This is the foundation of reusable, composable code.

**Concept Flow**

  1. return value   -->  sends value back to the caller
  2. result = f()  -->  captures the returned value
  3. print(f())    -->  uses the returned value directly
  4. Execution stops at return; code after return in the function doesn't run
  5. A function without return implicitly returns None

**Code Preview**

```python
# Adapted from source example 8.17
def build_profile(first, last, **user_info):
    """Build a dictionary containing everything we know about a user."""
    user_info['first_name'] = first
    user_info['last_name'] = last
    return user_info
user_profile = build_profile('albert', 'einstein',
                             location='princeton',
                             field='physics')
print(user_profile)
```

````python
# Adapted from source example 8.17
def build_profile(first, last, **user_info):
    """Build a dictionary containing everything we know about a user."""
    user_info['first_name'] = first
    user_info['last_name'] = last
    return user_info

user_profile = build_profile('albert', 'einstein',
                             location='princeton',
                             field='physics')
print(user_profile)
````

### Lab Exercise 8.21: Passing Lists to Functions

**Context**

When you pass a list to a function, the function receives a reference to the same list — it can read and modify it. This lets you process a whole collection with one function call: print all users, validate all inputs, format all names. Changes the function makes to the list are visible outside the function.

**Concept Flow**

  1. def f(my_list):  -->  receives the actual list (not a copy)
  2.     for item in my_list: ...  -->  iterate and process
  3. Modifications inside the function affect the original list
  4. To protect the original, pass a copy: f(my_list[:])

**Code Preview**

```python
# Adapted from source example 8.21
def greet_users(names):
    """Print a simple greeting to each user in the list."""
    for name in names:
        msg = f"Hello, {name.title()}!"
        print(msg)
usernames = ['hannah', 'ty', 'margot']
greet_users(usernames)
```

````python
# Adapted from source example 8.21
def greet_users(names):
    """Print a simple greeting to each user in the list."""
    for name in names:
        msg = f"Hello, {name.title()}!"
        print(msg)

usernames = ['hannah', 'ty', 'margot']
greet_users(usernames)
````

### Lab Exercise 8.26: Organising Functions in Modules and Importing Them

**Context**

A **module** is a `.py` file containing functions. Storing functions in modules means you can import them into any other program — your code becomes reusable across projects. `import module_name` imports the whole module; `from module_name import function` imports just one function. Aliases (`import module as m`) shorten long names.

**Concept Flow**

  1. import module_name               -->  access as module_name.function()
  2. from module import function      -->  call function() directly
  3. from module import function as f -->  alias: call f() instead
  4. import module as m               -->  alias the whole module: m.function()
  5. Keep import statements at the top of the file

**Code Preview**

```python
# Adapted from source example 8.26
import pizza
pizza.make_pizza(16, 'pepperoni')
pizza.make_pizza(12, 'mushrooms', 'green peppers', 'extra cheese')
```

````python
# Adapted from source example 8.26
import pizza


pizza.make_pizza(16, 'pepperoni')
pizza.make_pizza(12, 'mushrooms', 'green peppers', 'extra cheese')
````

### Lab Exercise 8.36: Returning Values from Functions

**Context**

A function can **return** a value to the calling code using `return`. The returned value can be stored in a variable, printed, or passed to another function. Return values let you separate *computation* (done inside the function) from *use* (done outside). This is the foundation of reusable, composable code.

**Concept Flow**

  1. return value   -->  sends value back to the caller
  2. result = f()  -->  captures the returned value
  3. print(f())    -->  uses the returned value directly
  4. Execution stops at return; code after return in the function doesn't run
  5. A function without return implicitly returns None

**Code Preview**

```python
# Adapted from source example 8.36
def build_person(first_name, last_name):
    """Return a dictionary of information about a person."""
    person = {'first': first_name, 'last': last_name}
    return person
musician = build_person('jimi', 'hendrix')
print(musician)
```

````python
# Adapted from source example 8.36
def build_person(first_name, last_name):
    """Return a dictionary of information about a person."""
    person = {'first': first_name, 'last': last_name}
    return person

musician = build_person('jimi', 'hendrix')
print(musician)
````

### Lab Exercise 8.48: Returning Values from Functions

**Context**

A function can **return** a value to the calling code using `return`. The returned value can be stored in a variable, printed, or passed to another function. Return values let you separate *computation* (done inside the function) from *use* (done outside). This is the foundation of reusable, composable code.

**Concept Flow**

  1. return value   -->  sends value back to the caller
  2. result = f()  -->  captures the returned value
  3. print(f())    -->  uses the returned value directly
  4. Execution stops at return; code after return in the function doesn't run
  5. A function without return implicitly returns None

**Code Preview**

```python
    return user_info
user_profile = build_profile('albert', 'einstein',
                             location='princeton',
                             field='physics')
print(user_profile)
```

### Lab Exercise 9.1: Modifying Attribute Values — Directly and Through Methods

**Context**

Instance attributes can be changed in three ways: directly (`instance.attr = value`), through a setter method (`instance.update(value)`), or through an increment method (`instance.increment(amount)`). Setter methods can include validation logic (e.g., refusing to roll back an odometer), which is why methods are often preferred over direct assignment.

**Concept Flow**

  1. instance.attr = value          -->  direct assignment (no validation)
  2. instance.update_attr(value)    -->  method can validate before setting
  3. instance.increment_attr(n)     -->  adds n to current value
  4. Methods with if-checks prevent invalid state (e.g., odometer can't go back)

**Code Preview**

```python
# Adapted from source example 9.1

    """A simple attempt to represent a car."""

        """Initialize attributes to describe a car."""

        self.model = model
        self.year = year
        self.odometer_reading = 0
    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()
    def read_odometer(self):
        """Print a statement showing the car's mileage."""
```

````python
# Adapted from source example 9.1
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        """
        Set the odometer reading to the given value.
        Reject the change if it attempts to roll the odometer back.
        """
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

    def increment_odometer(self, miles):
        """Add the given amount to the odometer reading."""
        self.odometer_reading += miles


my_used_car = Car('subaru', 'outback', 2019)
print(my_used_car.get_descriptive_name())

my_used_car.update_odometer(23_500)
my_used_car.read_odometer()

my_used_car.increment_odometer(100)
my_used_car.read_odometer()
````

### Lab Exercise 9.2: Creating a Class with __init__ and Methods

**Context**

A **class** is a blueprint for creating objects. The `__init__()` method runs automatically when a new instance is created; it sets up the instance's **attributes** using `self.attribute = value`. Methods are functions defined inside the class; they always take `self` as the first parameter, which gives them access to the instance's attributes.

Classes model real-world things: a `Dog` class represents any dog; a specific `Dog` instance represents your dog.

**Concept Flow**

  1. class MyClass:          -->  class definition (CapitalCase by convention)
  2.     def __init__(self, name):  -->  runs when MyClass('value') is called
  3.         self.name = name       -->  self.name is an instance attribute
  4.     def my_method(self):       -->  instance method
  5.         print(self.name)       -->  access attributes via self

**Code Preview**

```python
# Adapted from source example 9.2
class Dog:
    """A simple attempt to model a dog."""
    def __init__(self, name, age):
        """Initialize name and age attributes."""
        self.name = name
        self.age = age
    def sit(self):
        """Simulate a dog sitting in response to a command."""
        print(f"{self.name} is now sitting.")
    def roll_over(self):
        """Simulate rolling over in response to a command."""
        print(f"{self.name} rolled over!")
my_dog = Dog('Willie', 6)
your_dog = Dog('Lucy', 3)
```

````python
# Adapted from source example 9.2
class Dog:
    """A simple attempt to model a dog."""

    def __init__(self, name, age):
        """Initialize name and age attributes."""
        self.name = name
        self.age = age

    def sit(self):
        """Simulate a dog sitting in response to a command."""
        print(f"{self.name} is now sitting.")

    def roll_over(self):
        """Simulate rolling over in response to a command."""
        print(f"{self.name} rolled over!")


my_dog = Dog('Willie', 6)
your_dog = Dog('Lucy', 3)

print(f"My dog's name is {my_dog.name}.")
print(f"My dog is {my_dog.age} years old.")
my_dog.sit()

print(f"\nYour dog's name is {your_dog.name}.")
print(f"Your dog is {your_dog.age} years old.")
your_dog.sit()
````

### Lab Exercise 9.3: Using Instances as Attributes (Composition)

**Context**

Instead of cramming every feature into one class, you can break complex classes into smaller collaborating classes. `self.battery = Battery()` stores a `Battery` *instance* as an attribute of `ElectricCar`. You then access battery behaviour through the car: `my_car.battery.describe_battery()`. This is called **composition** and keeps each class focused and manageable.

**Concept Flow**

  1. self.component = ComponentClass()  -->  create and store as attribute
  2. instance.component.method()        -->  call method on the nested object
  3. instance.component.attribute       -->  access attribute of nested object
  4. Each class stays small; complexity is distributed across classes

**Code Preview**


````python
# Adapted from source example 9.3
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        """Set the odometer reading to the given value."""
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

    def increment_odometer(self, miles):
        """Add the given amount to the odometer reading."""
        self.odometer_reading += miles


class Battery:
    """A simple attempt to model a battery for an electric car."""

    def __init__(self, battery_size=40):
        """Initialize the battery's attributes."""
        self.battery_size = battery_size

    def describe_battery(self):
        """Print a statement describing the battery size."""
        print(f"This car has a {self.battery_size}-kWh battery.")

    def get_range(self):
        """Print a statement about the range this battery provides."""
        if self.battery_size == 40:
            range = 150
        elif self.battery_size == 65:
            range = 225

        print(f"This car can go about {range} miles on a full charge.")


class ElectricCar(Car):
    """Represent aspects of a car, specific to electric vehicles."""

    def __init__(self, make, model, year):
        """
        Initialize attributes of the parent class.
        Then initialize attributes specific to an electric car.
        """
        super().__init__(make, model, year)
        self.battery = Battery()


my_leaf = ElectricCar('nissan', 'leaf', 2024)
print(my_leaf.get_descriptive_name())
my_leaf.battery.describe_battery()
my_leaf.battery.get_range()
````

````python
# Adapted from source example 9.17
class Dog:
    """A simple attempt to model a dog."""

    def __init__(self, name, age):
        """Initialize name and age attributes."""
        self.name = name
        self.age = age

    def sit(self):
        """Simulate a dog sitting in response to a command."""
        print(f"{self.name} is now sitting.")

    def roll_over(self):
        """Simulate rolling over in response to a command."""
        print(f"{self.name} rolled over!")


my_dog = Dog('Willie', 6)
your_dog = Dog('Lucy', 3)

print(f"My dog's name is {my_dog.name}.")
print(f"My dog is {my_dog.age} years old.")
my_dog.sit()

print(f"\nYour dog's name is {your_dog.name}.")
print(f"Your dog is {your_dog.age} years old.")
your_dog.sit()
````

````python
# Adapted from source example 9.18
"""A class that can be used to represent a car."""

class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        """
        Set the odometer reading to the given value.
        Reject the change if it attempts to roll the odometer back.
        """
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

    def increment_odometer(self, miles):
        """Add the given amount to the odometer reading."""
        self.odometer_reading += miles
````

```python
# Adapted from source example 9.20
"""A set of classes used to represent gas and electric cars."""

class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
```
````python
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        """
        Set the odometer reading to the given value.
        Reject the change if it attempts to roll the odometer back.
        """
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

    def increment_odometer(self, miles):
        """Add the given amount to the odometer reading."""
        self.odometer_reading += miles


class Battery:
    """A simple attempt to model a battery for an electric car."""

    def __init__(self, battery_size=40):
        """Initialize the battery's attributes."""
        self.battery_size = battery_size

    def describe_battery(self):
        """Print a statement describing the battery size."""
        print(f"This car has a {self.battery_size}-kWh battery.")

    def get_range(self):
        """Print a statement about the range this battery provides."""
        if self.battery_size == 40:
            range = 150
        elif self.battery_size == 65:
            range = 225

        print(f"This car can go about {range} miles on a full charge.")

class ElectricCar(Car):
    """Models aspects of a car, specific to electric vehicles."""

    def __init__(self, make, model, year):
        """
        Initialize attributes of the parent class.
        Then initialize attributes specific to an electric car.
        """
        super().__init__(make, model, year)
        self.battery = Battery()
````

````python
# Adapted from source example 9.27
"""A set of classes that can be used to represent electric cars."""

from car import Car


class Battery:
    """A simple attempt to model a battery for an electric car."""

    def __init__(self, battery_size=40):
        """Initialize the battery's attributes."""
        self.battery_size = battery_size

    def describe_battery(self):
        """Print a statement describing the battery size."""
        print(f"This car has a {self.battery_size}-kWh battery.")

    def get_range(self):
        """Print a statement about the range this battery provides."""
        if self.battery_size == 40:
            range = 150
        elif self.battery_size == 65:
            range = 225

        print(f"This car can go about {range} miles on a full charge.")

class ElectricCar(Car):
    """Models aspects of a car, specific to electric vehicles."""

    def __init__(self, make, model, year):
        """
        Initialize attributes of the parent class.
        Then initialize attributes specific to an electric car.
        """
        super().__init__(make, model, year)
        self.battery = Battery()
````

### Lab Exercise 9.29: Modifying Attribute Values — Directly and Through Methods

**Context**

Instance attributes can be changed in three ways: directly (`instance.attr = value`), through a setter method (`instance.update(value)`), or through an increment method (`instance.increment(amount)`). Setter methods can include validation logic (e.g., refusing to roll back an odometer), which is why methods are often preferred over direct assignment.

**Concept Flow**

  1. instance.attr = value          -->  direct assignment (no validation)
  2. instance.update_attr(value)    -->  method can validate before setting
  3. instance.increment_attr(n)     -->  adds n to current value
  4. Methods with if-checks prevent invalid state (e.g., odometer can't go back)

**Code Preview**

```python
# Adapted from source example 9.29
class Car:
    """A simple attempt to represent a car."""
    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0
    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()
    def read_odometer(self):
        """Print a statement showing the car's mileage."""
```

````python
# Adapted from source example 9.29
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        """Set the odometer reading to the given value."""
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

    def increment_odometer(self, miles):
        """Add the given amount to the odometer reading."""
        self.odometer_reading += miles


class ElectricCar(Car):
    """Represent aspects of a car, specific to electric vehicles."""

    def __init__(self, make, model, year):
        """Initialize attributes of the parent class."""
        super().__init__(make, model, year)


my_leaf = ElectricCar('nissan', 'leaf', 2024)
print(my_leaf.get_descriptive_name())
````

### Lab Exercise 9.31: Using Instances as Attributes (Composition)

**Context**

Instead of cramming every feature into one class, you can break complex classes into smaller collaborating classes. `self.battery = Battery()` stores a `Battery` *instance* as an attribute of `ElectricCar`. You then access battery behaviour through the car: `my_car.battery.describe_battery()`. This is called **composition** and keeps each class focused and manageable.

**Concept Flow**

  1. self.component = ComponentClass()  -->  create and store as attribute
  2. instance.component.method()        -->  call method on the nested object
  3. instance.component.attribute       -->  access attribute of nested object
  4. Each class stays small; complexity is distributed across classes

**Code Preview**

```python
        self.odometer_reading = 0
    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()
    def read_odometer(self):
        """Print a statement showing the car's mileage."""
```

### Lab Exercise 9.37: Creating a Class with __init__ and Methods

**Context**

A **class** is a blueprint for creating objects. The `__init__()` method runs automatically when a new instance is created; it sets up the instance's **attributes** using `self.attribute = value`. Methods are functions defined inside the class; they always take `self` as the first parameter, which gives them access to the instance's attributes.

Classes model real-world things: a `Dog` class represents any dog; a specific `Dog` instance represents your dog.

**Concept Flow**

  1. class MyClass:          -->  class definition (CapitalCase by convention)
  2.     def __init__(self, name):  -->  runs when MyClass('value') is called
  3.         self.name = name       -->  self.name is an instance attribute
  4.     def my_method(self):       -->  instance method
  5.         print(self.name)       -->  access attributes via self

**Code Preview**

````python
# Adapted from source example 9.37
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        """
        Set the odometer reading to the given value.
        Reject the change if it attempts to roll the odometer back.
        """
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

my_new_car = Car('audi', 'a4', 2024)
print(my_new_car.get_descriptive_name())

my_new_car.update_odometer(23)
my_new_car.read_odometer()

my_new_car.update_odometer(10)
my_new_car.read_odometer()
````

### Lab Exercise 9.38: Modifying Attribute Values — Directly and Through Methods

**Context**

Instance attributes can be changed in three ways: directly (`instance.attr = value`), through a setter method (`instance.update(value)`), or through an increment method (`instance.increment(amount)`). Setter methods can include validation logic (e.g., refusing to roll back an odometer), which is why methods are often preferred over direct assignment.

**Concept Flow**

  1. instance.attr = value          -->  direct assignment (no validation)
  2. instance.update_attr(value)    -->  method can validate before setting
  3. instance.increment_attr(n)     -->  adds n to current value
  4. Methods with if-checks prevent invalid state (e.g., odometer can't go back)

**Code Preview**

```python
# Adapted from source example 9.38
class Car:
    """A simple attempt to represent a car."""
    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0
    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()
    def read_odometer(self):
        """Print a statement showing the car's mileage."""
```

````python
# Adapted from source example 9.38
class Car:
    """A simple attempt to represent a car."""

    def __init__(self, make, model, year):
        """Initialize attributes to describe a car."""
        self.make = make
        self.model = model
        self.year = year
        self.odometer_reading = 0

    def get_descriptive_name(self):
        """Return a neatly formatted descriptive name."""
        long_name = f"{self.year} {self.make} {self.model}"
        return long_name.title()

    def read_odometer(self):
        """Print a statement showing the car's mileage."""
        print(f"This car has {self.odometer_reading} miles on it.")

    def update_odometer(self, mileage):
        """
        Set the odometer reading to the given value.
        Reject the change if it attempts to roll the odometer back.
        """
        if mileage >= self.odometer_reading:
            self.odometer_reading = mileage
        else:
            print("You can't roll back an odometer!")

    def increment_odometer(self, miles):
        """Add the given amount to the odometer reading."""
        self.odometer_reading += miles


my_used_car = Car('subaru', 'outback', 2019)
print(my_used_car.get_descriptive_name())

my_used_car.update_odometer(23_500)
my_used_car.read_odometer()

my_used_car.increment_odometer(100)
my_used_car.read_odometer()
````

### Lab Exercise 10.1: Reading a File with Path and read_text()

**Context**

Python's `pathlib.Path` class makes file operations clean and cross-platform. `path.read_text()` reads the entire file into a string in one call. `rstrip()` removes the trailing newline that `read_text()` adds. This is the modern Python 3 way to read files — simpler and more readable than the older `open()` / `file.read()` pattern.

**Concept Flow**

  1. from pathlib import Path           -->  import the Path class
  2. path = Path('filename.txt')        -->  create a path object
  3. contents = path.read_text()        -->  read entire file as string
  4. contents = path.read_text().rstrip()  -->  strip trailing whitespace
  5. path.read_text(encoding='utf-8')   -->  specify encoding if needed

**Code Preview**

```python
from pathlib import Path
path = Path('alice.txt')
try:
    contents = path.read_text(encoding='utf-8')
except FileNotFoundError:
    print(f"Sorry, the file {path} does not exist.")
else:
    # Count the approximate number of words in the file:
    words = contents.split()
    num_words = len(words)
    print(f"The file {path} has about {num_words} words.")
```

````python
from pathlib import Path


path = Path('alice.txt')
try:
    contents = path.read_text(encoding='utf-8')
except FileNotFoundError:
    print(f"Sorry, the file {path} does not exist.")
else:
    # Count the approximate number of words in the file:
    words = contents.split()
    num_words = len(words)
    print(f"The file {path} has about {num_words} words.")
````

### Lab Exercise 10.6: Failing Silently with pass in the except Block

**Context**

Sometimes you don't want to tell users about every error — you just want to skip the problem and continue. Use `pass` in the `except` block to do nothing: no message, no traceback, just move on. The book notes that `pass` also serves as a placeholder — a visual reminder that you *chose* to do nothing there, not that you forgot.

**Concept Flow**

  1. except FileNotFoundError:
  2.     pass          -->  do nothing; continue silently
  3. Users see normal results; missing/broken files are skipped silently
  4. pass is also useful as a placeholder for future code

**Code Preview**

```python
from pathlib import Path
def count_words(path):
    """Count the approximate number of words in a file."""
    try:
        contents = path.read_text(encoding='utf-8')
    except FileNotFoundError:
        pass
    else:
        # Count the approximate number of words in the file:
        words = contents.split()
        num_words = len(words)
        print(f"The file {path} has about {num_words} words.")
# ... 6 more line(s) in this source example
```

### Lab Exercise 10.32: Failing Silently with pass in the except Block

**Context**

Sometimes you don't want to tell users about every error — you just want to skip the problem and continue. Use `pass` in the `except` block to do nothing: no message, no traceback, just move on. The book notes that `pass` also serves as a placeholder — a visual reminder that you *chose* to do nothing there, not that you forgot.

**Concept Flow**

  1. except FileNotFoundError:
  2.     pass          -->  do nothing; continue silently
  3. Users see normal results; missing/broken files are skipped silently
  4. pass is also useful as a placeholder for future code

**Code Preview**

```python
from pathlib import Path
def count_words(path):
    """Count the approximate number of words in a file."""
    try:

    except FileNotFoundError:
        pass
    else:
        # Count the approximate number of words in the file:
        words = contents.split()
        num_words = len(words)
        print(f"The file {path} has about {num_words} words.")
# ... 6 more line(s) in this source example
```

### Lab Exercise 10.48: Remembering User Data Between Runs with json

**Context**

Combining `json`, `Path.exists()`, and `if-else` lets you build a program that remembers information between runs. Check if the data file exists; if yes, load and greet the user; if no, ask for their name and save it. This is the foundation of persistent applications — preferences, save games, cached results.

**Concept Flow**

  1. path.exists()  -->  True if file exists, False otherwise
  2. if path.exists(): load and use  else: prompt and save
  3. json.loads(path.read_text())   -->  reload saved data
  4. json.dumps(data); path.write_text(...)  -->  save new data

**Code Preview**

```python
from pathlib import Path
import json
path = Path('username.json')
if path.exists():
    contents = path.read_text()
    username = json.loads(contents)
    print(f"Welcome back, {username}!")
else:
    username = input("What is your name? ")
    contents = json.dumps(username)
    path.write_text(contents)
    print(f"We'll remember you when you come back, {username}!")
```

````python
from pathlib import Path
import json


path = Path('username.json')
if path.exists():
    contents = path.read_text()
    username = json.loads(contents)
    print(f"Welcome back, {username}!")
else:
    username = input("What is your name? ")
    contents = json.dumps(username)
    path.write_text(contents)
    print(f"We'll remember you when you come back, {username}!")
````

**Context**

This is how you persist program state between runs and how you exchange data with web APIs.

**Concept Flow**

  1. import json
  2. json.dumps(data)          -->  Python object → JSON string
  3. json.loads(string)        -->  JSON string → Python object
  4. path.write_text(json.dumps(data))  -->  save to file
  5. json.loads(path.read_text())       -->  load from file

### Lab Exercise 11.1: Writing and Running Unit Tests with pytest

**Context**

A **unit test** verifies that one specific behaviour of a function is correct. You write test functions (names starting with `test_`) that call your function and then **assert** something about the result. `pytest` discovers and runs all `test_` files and reports which tests pass or fail. This gives you confidence that your code works — and alerts you immediately when changes break something.

**Concept Flow**

  1. test functions must start with test_
  2. assert result == expected   -->  passes if equal; fails with details if not
  3. pytest discovers test_ files and test_ functions automatically
  4. Passing test: dot (.) in output. Failing test: F with error details
  5. Run with: pytest  or  python -m pytest

````python
# Adapted from source example 11.4
class AnonymousSurvey:
    """Collect anonymous answers to a survey question."""

    def __init__(self, question):
        """Store a question, and prepare to store responses."""
        self.question = question
        self.responses = []

    def show_question(self):
        """Show the survey question."""
        print(self.question)

    def store_response(self, new_response):
        """Store a single response to the survey."""
        self.responses.append(new_response)

    def show_results(self):
        """Show all the responses that have been given."""
        print("Survey results:")
        for response in self.responses:
            print(f"- {response}")
````

### Lab Exercise 11.6: Reducing Repetition in Tests with pytest Fixtures

**Context**

A **fixture** is a function marked with `@pytest.fixture` that creates a shared resource used by multiple tests. Instead of creating a new class instance in every test function, the fixture creates it once and pytest injects it automatically as a parameter. This keeps tests DRY and makes the setup code easy to update in one place.

**Concept Flow**

  1. @pytest.fixture            -->  marks a function as a fixture
  2. def my_fixture(): ...      -->  creates and returns the shared resource
  3. def test_something(my_fixture):  -->  pytest injects the fixture
  4. fixture runs before each test function that requests it
  5. Fixtures can also do teardown (cleanup) after the test

**Code Preview**

```python
import pytest
from survey import AnonymousSurvey
@pytest.fixture
def language_survey():
    """A survey that will be available to all test functions."""
    question = "What language did you first learn to speak?"
    language_survey = AnonymousSurvey(question)
    return language_survey
def test_store_single_response(language_survey):
    """Test that a single response is stored properly."""
    language_survey.store_response('English')
# ... 10 more line(s) in this source example
```
