## Lab 4.7 — Copying a List
---
### Context
`list_b = list_a` does **not** create a copy — both variables share the same list in memory. Use `[:]` to create a true independent copy. This distinction is critical whenever you need to preserve original data.

---
### Concept Flow

```mermaid
flowchart TD
    subgraph WRONG["❌ Wrong — Shared Reference (same memory)"]
        direction LR
        A1["my_foods = ['pizza','falafel']"]:::list
        B1["friend_foods = my_foods\nSame memory address!"]:::shared
        C1["my_foods.append('cannoli')\n──▶ BOTH lists get 'cannoli'"]:::fx
        A1 --> B1 --> C1
    end
    subgraph RIGHT["✅ Correct — True Copy with [:]"]
        direction LR
        A2["my_foods = ['pizza','falafel']"]:::list
        B2["friend_foods = my_foods[:]\nNew independent list!"]:::copy
        C2["my_foods.append('cannoli')\nfriend_foods.append('ice cream')"]:::ind
        D2["my_foods:     ['pizza','falafel','cannoli']\nfriend_foods: ['pizza','falafel','ice cream']"]:::res
        A2 --> B2 --> C2 --> D2
    end
    classDef list   fill:#dce8f5,stroke:#5b9bd5,color:#1a1a2e,font-weight:bold
    classDef shared fill:#f8d7da,stroke:#c0392b,color:#1a1a2e,font-weight:bold
    classDef fx     fill:#f8d7da,stroke:#c0392b,color:#1a1a2e
    classDef copy   fill:#e8f4e8,stroke:#5ba85b,color:#1a1a2e,font-weight:bold
    classDef ind    fill:#e8f4e8,stroke:#5ba85b,color:#1a1a2e
    classDef res    fill:#f3e8ff,stroke:#9b5bd5,color:#1a1a2e,font-weight:bold
```

### Code Preview
