## Lab 5.1 -- Conditional Tests
### `conditional_tests.py`
---
### Context
Before writing any `if` statement you need to understand **conditional tests** -- expressions that evaluate to `True` or `False`. Python uses `==` (equality), `!=` (inequality), `<`, `>`, `<=`, `>=` for comparisons, and `and`/`or` to combine them. These are the raw building blocks of every decision your program makes.

---
### Code Preview
```python
car = 'bmw'
print(car == 'bmw')       # True  -- exact match
print(car == 'audi')      # False -- no match

# Case-insensitive comparison
car = 'Audi'
print(car.lower() == 'audi')    # True

# Inequality
print(car.lower() != 'bmw')     # True

# Numerical comparisons
age = 19
print(age < 21)   # True
print(age <= 21)  # True
print(age > 21)   # False
print(age >= 21)  # False
print(age == 21)  # False

# and / or
age_0, age_1 = 22, 18
print(age_0 >= 21 and age_1 >= 21)   # False
print(age_0 >= 21 or  age_1 >= 21)   # True

# in / not in
toppings = ['mushrooms', 'onions', 'pineapple']
print('mushrooms' in toppings)     # True
print('pepperoni' in toppings)     # False

# Boolean values
game_active = True
can_edit    = False
print(game_active, can_edit)
```

## Lab 6.3 -- Using get() for Safe Access
### `alien.py` (part 3)
---
### Context
Accessing a key that doesn't exist raises a `KeyError`. The `.get(key, default)` method avoids this by returning a default value if the key is missing. Use `get()` whenever you're not sure a key will be present -- reading user input, parsing API responses, etc.

---
### Code Preview
```python
alien_0 = {'color': 'green', 'speed': 'slow'}

# Safe access with get()
point_value = alien_0.get('points', 'No point value assigned.')
print(point_value)   # No point value assigned.

# If key exists, get() returns its value
color = alien_0.get('color', 'No color assigned.')
print(color)   # green

# Dangerous (raises KeyError) -- DON'T do this if key might be missing:
# print(alien_0['points'])
```

## Lab 6.5 -- Looping Through Keys
### `favorite_languages.py` (part 2)
---
### Context
`.keys()` returns only the keys. Since looping over a dict already gives keys by default, `.keys()` mainly adds readability -- or lets you convert keys to a `list()`, `set()`, or `sorted()` sequence. Sorting is especially useful when order matters for output.

### Code Preview
```python
favorite_languages = {
    'jen':   'python',
    'sarah': 'c',
    'edward':'ruby',
    'phil':  'python',

# Default loop -- same as .keys()
for name in favorite_languages:
    print(name.title())

# Explicitly using .keys()
for name in favorite_languages.keys():
    print(name.title())

# Check if a specific key exists
if 'erin' not in favorite_languages.keys():
    print("Erin, please take our poll!")

# Sorted keys
for name in sorted(favorite_languages.keys()):
    print(f"{name.title()}, thank you for taking the poll.")
```

## Lab 6.6 -- Looping Through Values / Unique Values with set()
### `favorite_languages.py` (part 3)
---
### Context
`.values()` returns only the values. Wrapping in `set()` removes duplicates, giving you the unique collection. This is handy for reporting ("which languages are popular?") without counting individual people.

---
### Code Preview
```python

    'jen':   'python',
    'sarah': 'c',
    'edward':'ruby',
    'phil':  'python',
}

print("The following languages have been mentioned:")
for language in favorite_languages.values():
    print(language.title())

# Unique values only
print("
Unique languages mentioned:")
for language in set(favorite_languages.values()):
    print(language.title())

# Sorted unique values
for language in sorted(set(favorite_languages.values())):
```

## Lab 10.9 -- Storing Data with JSON
### `number_writer.py` / `number_reader.py`
---
### Context
JSON (JavaScript Object Notation) is a text format for storing Python dicts, lists, strings, and numbers. `json.dumps(data)` converts Python to a JSON string. `json.loads(text)` parses it back. `Path.write_text()` / `read_text()` do the file I/O. This is how programs save and reload state.

---
### Code Preview
```python
from pathlib import Path
import json

contents = json.dumps(numbers)
path.write_text(contents)
print(f"Saved: {contents}")

# Read them back
path     = Path('numbers.json')
contents = path.read_text()
numbers  = json.loads(contents)
print(numbers)
print(type(numbers))   # <class 'list'>
```

```python
import json
```
